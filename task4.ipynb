task4. Simple multi-class classification using neural networks.

```python
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.datasets import make_blobs
import tensorflow as tf 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
#I will use the 'make_blobs' function to make a dataset with 4 categories
X_train, y_train = make_blobs(n_samples=500,centers=4,cluster_std=1.5,random_state=42,n_features=2)

plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap='viridis', alpha=0.7)
plt.xlabel("Feature-1")
plt.ylabel("Feature-2")
plt.show()
```

```python
#show classes in data set
print(f"unique classes {np.unique(y_train)}")

#show how classes are represented
print(f"class representation {y_train[:10]}")

#show shapes
print(f"shapes {X_train.shape}, {y_train.shape}")
```

```python
tf.random.set_seed(1234)
model = Sequential(
    [
        Dense(2,activation='relu',name="L1"),
        Dense(4,activation='linear',name="L2")
    ]
)
```

```python
model.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer=tf.keras.optimizers.Adam(0.01),
)

model.fit(X_train,y_train,epochs=200)
```

```python
#gathering the training parameters from layer 1
l1 = model.get_layer("L1")
W1,b1 = l1.get_weights()

print(f"W1: {W1}")
print(f"b1: {b1}")
```
